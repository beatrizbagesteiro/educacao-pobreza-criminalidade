```python
import pandas as pd
from pathlib import Path
```

```python
root = Path.cwd()
data_path = root / 'work' / 'data' / 'raw' / 'cad_unico_2024.csv'
```

```python
df_cad_unico = pd.read_csv(data_path, encoding= 'utf-8')
df_cad_unico.head()
```

```python
df_cad_unico = df_cad_unico.drop(columns=['Unidade Territorial', 'UF', 'Referência'])
```

```python
df_cad_unico = df_cad_unico.rename(columns={
    "Código": "co_municipio", 
    "Referência": "referencia",
    "Número de famílias cadastradas no Cadastro Único em situação de pobreza, segundo a faixa do Programa Bolsa Família*": "familias_pobreza",
    "Número de famílias cadastradas no Cadastro Único com renda per capita de até meio salário-mínimo (Pobreza + Baixa renda)": "familias_baixa_renda",
    "Número de famílias cadastradas no Cadastro Único com renda per capita acima de meio salário-mínimo***": "familias_acima_meio_sm"
})

df_cad_unico.head()
```

```python
df_cad_unico.info()
```

```python
df_cad_unico = df_cad_unico.groupby('co_municipio').aggregate('mean').round()
df_cad_unico.head()
```
