```python
import pandas as pd
```

```python
from pathlib import Path
root =  Path.cwd().parent 
data_path = root / 'data' / 'raw' / 'censo_escolar_2024.csv'
print(data_path)
```

```python
df_censo_escolar = pd.read_csv(data_path, encoding='latin1', delimiter=';')
df_censo_escolar.head()
```

```python
colunas = list(df_censo_escolar.columns)
colunas
```

```python
df_censo_escolar.info(verbose=True)
```

```python
df_censo_escolar = df_censo_escolar[df_censo_escolar['TP_SITUACAO_FUNCIONAMENTO'] == 1]
df_censo_escolar.info()
```

```python
prefixos_infra = ('IN_AGUA', 'IN_ENERGIA', 'IN_ESGOTO', 'IN_LIXO', 'IN_INTERNET', 'IN_BANHEIRO', 'IN_ACESSABILIDADE', 'IN_COZINHA', 'IN_REFEITORIO', 'IN_BIBLIOTECA', 'IN_QUADRA_ESPORTES', 'IN_ALIMENTACAO', 'IN_LABORATORIO')
colunas = df_censo_escolar.columns

colunas_infra = []
for coluna in colunas:
    if coluna.startswith(prefixos_infra):
        colunas_infra.append(coluna)

colunas_final = ['NO_REGIAO','CO_MUNICIPIO', 'NO_MUNICIPIO', 'SG_UF'] + colunas_infra
df_censo_escolar = df_censo_escolar.filter(colunas_final)
df_censo_escolar.head()
```

```python
df_censo_escolar = df_censo_escolar[df_censo_escolar['co_municipio'] == 1]
```

```python
df_censo_escolar['co_municipio'] = df_censo_escolar['co_municipio'].astype(str).str[:-1]
```

```python
df_censo_escolar.drop_duplicates('NO_REGIAO')
```

```python
df_censo_escolar.columns = df_censo_escolar.columns.str.lower()
df_censo_escolar.info()
```

```python
total_nulos = df_censo_escolar.isnull().sum().sum()
print(total_nulos)
```

```python
colunas_num = df_censo_escolar.select_dtypes(include='number')
valores_nao_bool = ((colunas_num < 0) | (colunas_num > 1)).sum()
print(valores_nao_bool)
```
