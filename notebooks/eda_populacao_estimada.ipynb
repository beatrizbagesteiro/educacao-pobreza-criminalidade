```python
import pandas as pd
from pathlib import Path
```

```python
root = Path.cwd().parent
data_dir = root / 'data' / 'raw' / 'populacao_estimada_2024.json'
df_populacao = pd.read_json(data_dir)
df_populacao
```

```python
df_populacao = df_populacao.drop(columns=['NC','NN','MC','MN','D1C','D1N','D2C','D2N'])
df_populacao
```

```python
df_populacao = df_populacao.rename(columns={'V':'cont_populacao', 'D3C': 'co_municipio', 'D3N':'no_municipio'})
df_populacao
```

```python
df_populacao = df_populacao.iloc[1:].reset_index(drop=True)
df_populacao
```

```python
df_populacao['co_municipio'] = df_populacao['co_municipio'].astype(str).str[:-1]
```

```python
total_nulos = df_populacao.isnull().sum().sum()
print(total_nulos)
```

```python
linhas_duplicadas = df_populacao.duplicated(keep='first').sum()
linhas_duplicadas
```

```python
linhas_duplicadas = df_populacao.duplicated(subset=['co_municipio'], keep='first').sum()
print(linhas_duplicadas)
```
