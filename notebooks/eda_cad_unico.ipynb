```python
import pandas as pd
import seaborn as sns
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
root = Path.cwd().parent
data_path = root / 'data' / 'raw' / 'cad_unico_2024.csv'
```

```python
df_cad_unico = pd.read_csv(data_path, encoding= 'latin-1')
df_cad_unico.head()
```

```python
df_cad_unico.info()
```

```python
df_cad_unico = df_cad_unico.rename(columns={
    "Código": "co_municipio", 
    "Unidade Territorial": "no_municipio",
    "UF": "sg_uf",
    "Referência": "mes_referencia",
    "Número de famílias cadastradas no Cadastro Único em situação de pobreza, segundo a faixa do Programa Bolsa Família*": "familias_pobreza",
    "Número de famílias cadastradas no Cadastro Único com renda per capita de até meio salário-mínimo (Pobreza + Baixa renda)": "familias_baixa_renda",
    "Número de famílias cadastradas no Cadastro Único com renda per capita acima de meio salário-mínimo***": "familias_acima_meio_sm"
})

df_cad_unico.head()
```

```python
df_cad_unico['mes_referencia'] = pd.to_datetime(df_cad_unico['mes_referencia'], format='%m/%Y')
df_cad_unico
```

```python
# Não há valores duplicados
'''
keep : {first, last, False}, default first

first : Mark duplicates as True except for the first occurrence.
last : Mark duplicates as True except for the last occurrence.
False : Mark all duplicates as True.
'''
linhas_duplicadas = df_cad_unico.duplicated(keep='first').sum()
linhas_duplicadas
```

```python
# Não há valores nulos
total_nulos = df_cad_unico.isnull().sum().sum()
print(total_nulos)
```

```python
df_cidade = df_cad_unico[df_cad_unico['no_municipio'] == 'ACRELÂNDIA'].sort_values('mes_referencia')
print(df_cidade)
```
